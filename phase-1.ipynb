## 1. The NumPy Advantage: Speed & Vectorization
Why is NumPy the universal standard for scientific computing?

- **Contiguous Memory:** NumPy stores data in continuous memory blocks for optimal CPU cache utilization.
- **SIMD Vectorization:** Operations execute through compiled C/Fortran vector instructions.
- **Homogeneous Data Types:** No dynamic type overhead per element.

```python
import numpy as np
import time

# Create a Python list and a NumPy array with 1,000,000 numbers
size = 1_000_000
py_list = list(range(size))
np_arr = np.arange(size)

# Benchmark Python list: element-wise multiplication by 2
start_time = time.perf_counter()
py_result = [x * 2 for x in py_list]
py_duration = time.perf_counter() - start_time

# Benchmark NumPy array: vectorized multiplication by 2
start_time = time.perf_counter()
np_result = np_arr * 2
np_duration = time.perf_counter() - start_time

print(f"Python list comprehension time : {py_duration:.5f} seconds")
print(f"NumPy vectorized operation time : {np_duration:.5f} seconds")
print(f"🚀 NumPy is approximately {py_duration / np_duration:.1f}x faster!")
```

### Element-wise Arithmetic vs. List Repetition
A classic beginner pitfall is expecting Python lists to behave mathematically:

```python
# Python lists repeat when multiplied:
py_sample = [1, 2, 3]
print("Python list * 2:", py_sample * 2)

# NumPy arrays perform mathematical element-wise multiplication:
np_sample = np.array([1, 2, 3])
print("NumPy array * 2 :", np_sample * 2)
```

## 2. Creating Arrays from Python Collections
NumPy structures data into multi-dimensional arrays (`ndarray`):

```python
# 1D Array (Vector)
vector = np.array([10, 20, 30, 40, 50])
print("1D Vector:\n", vector)

# 2D Array (Matrix: rows x columns)
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print("\n2D Matrix:\n", matrix)

# 3D Array (Tensor: depth x rows x columns)
tensor = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print("\n3D Tensor:\n", tensor)
```

## 3. Intrinsic Array Generators
NumPy offers high-speed generator functions for common patterns:

```python
# Constant filled arrays
zeros = np.zeros((2, 4))               # 2 rows, 4 columns of 0.0
ones = np.ones((3, 2))                 # 3 rows, 2 columns of 1.0
full = np.full((2, 3), fill_value=7)   # 2 rows, 3 columns filled with 7
identity = np.eye(3)                   # 3x3 Identity matrix

print("np.zeros((2, 4)):\n", zeros)
print("\nnp.ones((3, 2)):\n", ones)
print("\nnp.full((2, 3), 7):\n", full)
print("\nnp.eye(3):\n", identity)
```

### Numerical Sequences: `np.arange()` vs `np.linspace()`
- Use **`np.arange(start, stop, step)`** when you know the step size.
- Use **`np.linspace(start, stop, count)`** when you know the exact number of data points needed.

```python
# arange: step = 2 (stop is exclusive)
range_arr = np.arange(0, 11, 2)
print("np.arange(0, 11, 2):", range_arr)

# linspace: exactly 5 evenly-spaced points between 0 and 1 (stop is inclusive)
linspace_arr = np.linspace(0, 1, 5)
print("np.linspace(0, 1, 5):", linspace_arr)
```

### Reproducible Random Number Generation
Setting `np.random.seed()` ensures reproducible data across experiments:

```python
np.random.seed(42)

# Uniform random numbers in [0.0, 1.0)
rand_uniform = np.random.rand(2, 3)

# Standard normal distribution (mean=0, variance=1)
rand_normal = np.random.randn(2, 3)

# Random integers between [10, 100)
rand_ints = np.random.randint(10, 100, size=(3, 3))

print("Uniform Random:\n", rand_uniform)
print("\nNormal Random:\n", rand_normal)
print("\nRandom Integers:\n", rand_ints)
```

## 4. Inspecting Array Anatomy
Key attributes to inspect any NumPy array:

```python
arr_demo = np.array([[1.5, 2.5, 3.5], [4.5, 5.5, 6.5]])

print("Array:\n", arr_demo)
print(f"Shape (rows, cols)     : {arr_demo.shape}")
print(f"Dimensions (.ndim)     : {arr_demo.ndim}")
print(f"Total elements (.size) : {arr_demo.size}")
print(f"Data type (.dtype)     : {arr_demo.dtype}")
print(f"Bytes per element      : {arr_demo.itemsize} bytes")
print(f"Total memory consumed  : {arr_demo.nbytes} bytes")
```

### Memory Optimization via Type Casting (`.astype()`)
Choosing the right dtype drastically reduces memory requirements:

```python
float_arr = np.array([10.2, 20.7, 30.1, 40.8])
int_arr = float_arr.astype(np.int32)

print(f"Original float64 memory : {float_arr.nbytes} bytes")
print(f"Downcasted int32 memory : {int_arr.nbytes} bytes (50% reduction!)")
```

## 5. Array Reshaping & Restructuring
Reshaping alters dimensions without copying data buffer.

> **Rule:** The total element count must remain identical (`rows * cols == total elements`).
> Pass `-1` to let NumPy calculate the dimension automatically.

```python
arr = np.arange(12)
print("Original 1D array:", arr)

# Reshape into 3 rows, 4 columns
matrix_3x4 = arr.reshape(3, 4)
print("\nReshaped to (3, 4):\n", matrix_3x4)

# Automatic dimension deduction with -1:
matrix_2x6 = arr.reshape(2, -1)
print("\nReshaped to (2, -1) -> (2, 6):\n", matrix_2x6)

# Matrix Transpose (swap rows and columns)
transposed = matrix_3x4.T
print("\nTransposed (4, 3):\n", transposed)
```

### `.flatten()` vs. `.ravel()` (Copy vs. View)
- **`.flatten()`**: Always allocates an independent copy.
- **`.ravel()`**: Returns a memory view whenever possible (much faster).

```python
flattened = matrix_3x4.flatten()
raveled = matrix_3x4.ravel()

# Mutating the raveled view mutates the original matrix:
raveled[0] = 999
print("Mutated raveled view -> matrix_3x4[0, 0] is now:", matrix_3x4[0, 0])
print("Flattened copy remained independent           :", flattened[0])
```

## 6. Hands-On Practice Challenge
**Task:**
1. Create an 8x8 matrix filled with zeros (`dtype=np.int32`).
2. Set the entire outer border (first/last row, first/last column) to `1`.
3. Check its `.shape` and verify memory footprint.

*Try coding your solution before checking below!*

```python
# Solution:
border_matrix = np.zeros((8, 8), dtype=np.int32)
border_matrix[0, :] = 1    # Top row
border_matrix[-1, :] = 1   # Bottom row
border_matrix[:, 0] = 1    # Left column
border_matrix[:, -1] = 1   # Right column

print("8x8 Border Matrix:\n", border_matrix)
print(f"Shape: {border_matrix.shape}, Memory: {border_matrix.nbytes} bytes")
```
